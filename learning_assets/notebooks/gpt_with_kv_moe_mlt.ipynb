```python
#####################################
# NEW: GQA instead of MHA
#####################################
class GroupedQueryAttention(nn.Module):
    def __init__(
            self, d_in, d_out, dropout, num_heads, num_kv_groups, dtype=None, qkv_bias=False
    ):
        super().__init__()
        assert d_out % num_heads == 0, "d_out must be divisible by num_heads"
        assert num_heads % num_kv_groups == 0, "num_heads must be divisible by num_kv_groups"

        self.d_out = d_out
        self.num_heads = num_heads
        self.head_dim = d_out // num_heads

        self.W_key = nn.Linear(d_in, num_kv_groups * self.head_dim, bias=qkv_bias, dtype=dtype)
        self.W_value = nn.Linear(d_in, num_kv_groups * self.head_dim, bias=qkv_bias, dtype=dtype)
        self.num_kv_groups = num_kv_groups
        self.group_size = num_heads // num_kv_groups

        self.W_query = nn.Linear(d_in, d_out, bias=qkv_bias, dtype=dtype)
        self.out_proj = nn.Linear(d_out, d_out, bias=False, dtype=dtype)
        self.dropout = nn.Dropout(dropout)

        self.register_buffer("cache_k", None, persistent=False)
        self.register_buffer("cache_v", None, persistent=False)
        self.ptr_current_pos = 0

    def forward(self, x, use_cache=False):
        b, num_tokens, _ = x.shape

        # Apply projections
        queries = self.W_query(x)  # (b, num_tokens, num_heads * head_dim)
        keys = self.W_key(x)       # (b, num_tokens, num_kv_groups * head_dim)
        values = self.W_value(x)   # (b, num_tokens, num_kv_groups * head_dim)

        # Reshape
        queries = queries.view(b, num_tokens, self.num_heads, self.head_dim).transpose(1, 2)
        keys_new = keys.view(b, num_tokens, self.num_kv_groups, self.head_dim).transpose(1, 2)
        values_new = values.view(b, num_tokens, self.num_kv_groups, self.head_dim).transpose(1, 2)

        if use_cache:
            if self.cache_k is None:
                self.cache_k, self.cache_v = keys_new, values_new
            else:
                self.cache_k = torch.cat([self.cache_k, keys_new], dim=2)
                self.cache_v = torch.cat([self.cache_v, values_new], dim=2)
            keys_base, values_base = self.cache_k, self.cache_v
        else:
            keys_base, values_base = keys_new, values_new
            if self.cache_k is not None or self.cache_v is not None:
                self.cache_k, self.cache_v = None, None
                self.ptr_current_pos = 0

        # Expand keys and values to match the number of heads
        # Shape: (b, num_heads, num_tokens, head_dim)
        keys = keys_base.repeat_interleave(self.group_size, dim=1)  # Shape: (b, num_heads, num_tokens, head_dim)
        values = values_base.repeat_interleave(self.group_size, dim=1)  # Shape: (b, num_heads, num_tokens, head_dim)
        # For example, before repeat_interleave along dim=1 (query groups):
        #   [K1, K2]
        # After repeat_interleave (each query group is repeated group_size times):
        #   [K1, K1, K2, K2]
        # If we used regular repeat instead of repeat_interleave, we'd get:
        #   [K1, K2, K1, K2]

        # Compute scaled dot-product attention (aka self-attention) with a causal mask
        # Shape: (b, num_heads, num_tokens, num_tokens)
        attn_scores = queries @ keys.transpose(2, 3)  # Dot product for each head

        ####################################################
        # causal mask
        num_tokens_Q = queries.shape[-2]
        num_tokens_K = keys.shape[-2]
        device = queries.device
        if use_cache:
            q_positions = torch.arange(
                self.ptr_current_pos,
                self.ptr_current_pos + num_tokens_Q,
                device=device,
                dtype=torch.long,
            )
            self.ptr_current_pos += num_tokens_Q
        else:
            q_positions = torch.arange(num_tokens_Q, device=device, dtype=torch.long)
            self.ptr_current_pos = 0
        k_positions = torch.arange(num_tokens_K, device=device, dtype=torch.long)
        mask = q_positions.unsqueeze(-1) < k_positions.unsqueeze(0)

        # Use the mask to fill attention scores
        attn_scores = attn_scores.masked_fill(mask, -torch.inf)

        attn_weights = torch.softmax(attn_scores / keys.shape[-1]**0.5, dim=-1)
        assert keys.shape[-1] == self.head_dim
        attn_weights = self.dropout(attn_weights)

        # Shape: (b, num_tokens, num_heads, head_dim)
        context_vec = (attn_weights @ values).transpose(1, 2)

        # Combine heads, where self.d_out = self.num_heads * self.head_dim
        context_vec = context_vec.contiguous().view(b, num_tokens, self.d_out)
        context_vec = self.out_proj(context_vec)  # optional projection

        return context_vec

    def reset_cache(self):
        self.cache_k, self.cache_v = None, None
        self.ptr_current_pos = 0
```

```python
import torch
import torch.nn as nn
#####################################
# Chapter 4
#####################################
class LayerNorm(nn.Module):
    def __init__(self, emb_dim):
        super().__init__()
        self.eps = 1e-5
        self.scale = nn.Parameter(torch.ones(emb_dim))
        self.shift = nn.Parameter(torch.zeros(emb_dim))

    def forward(self, x):
        mean = x.mean(dim=-1, keepdim=True)
        var = x.var(dim=-1, keepdim=True, unbiased=False)
        norm_x = (x - mean) / torch.sqrt(var + self.eps)
        return self.scale * norm_x + self.shift


class RMSNorm(nn.Module):
    def __init__(self, emb_dim, eps=1e-5):
        super().__init__()
        self.eps = eps
        self.emb_dim = emb_dim
        self.weight = nn.Parameter(torch.ones(emb_dim)).float()

    def forward(self, x):
        means = x.pow(2).mean(dim=-1, keepdim=True)
        x_normed = x * torch.rsqrt(means + self.eps)
        return (x_normed * self.weight).to(dtype=x.dtype)
```

```python
class GELU(nn.Module):
    def __init__(self):
        super().__init__()

    def forward(self, x):
        return 0.5 * x * (1 + torch.tanh(
            torch.sqrt(torch.tensor(2.0 / torch.pi)) *
            (x + 0.044715 * torch.pow(x, 3))
        ))

# class FeedForward(nn.Module):
#     def __init__(self, cfg):
#         super().__init__()
#         self.layers = nn.Sequential(
#             nn.Linear(cfg["emb_dim"], 4 * cfg["emb_dim"]),
#             GELU(),
#             nn.Linear(4 * cfg["emb_dim"], cfg["emb_dim"]),
#         )

#     def forward(self, x):
#         return self.layers(x)
```

```python
class SiLU(nn.Module):
    def __init__(self):
        super(SiLU, self).__init__()

    def forward(self, x):
        return x * torch.sigmoid(x)
silu = SiLU()

assert torch.allclose(silu(example_batch), torch.nn.functional.silu(example_batch))


# Uses SwiGLU instead of GeLU to make it more comparable to MoE
class FeedForward(nn.Module):
    def __init__(self, cfg):
        super().__init__()
        self.gate_proj = nn.Linear(cfg["emb_dim"], cfg["hidden_dim"], dtype=cfg["dtype"], bias=False)
        self.value_proj = nn.Linear(cfg["emb_dim"], cfg["hidden_dim"], dtype=cfg["dtype"], bias=False)
        self.out_proj = nn.Linear(cfg["hidden_dim"], cfg["emb_dim"], dtype=cfg["dtype"], bias=False)
        self.silu = SiLU()

    def forward(self, x):
        x_gate = self.gate_proj(x)
        x_value = self.value_proj(x)
        x = self.silu(x_gate) * x_value
        return self.out_proj(x)
```

```python
class MoEFeedForward(nn.Module):
    def __init__(self, cfg):
        super().__init__()

        self.num_experts = cfg["num_experts"]
        self.num_experts_per_tok = cfg["num_experts_per_tok"]
        self.emb_dim = cfg["emb_dim"]
        self.hidden_dim = cfg["hidden_dim"]

        # ============================================================
        # ROUTER
        #
        # x: (B, T, emb_dim)
        #
        # scores:
        # (B, T, num_experts)
        #
        # Mỗi token có 1 score cho từng expert.
        # ============================================================
        self.gate = nn.Linear(
            self.emb_dim,
            self.num_experts,
            bias=False
        )

        # Routing bias:
        #
        # Chỉ dùng để quyết định expert nào được chọn.
        # KHÔNG dùng làm routing weight cuối.
        #
        # DeepSeek cũng tách:
        #   selection score = score + bias
        #   routing weight  = original score
        self.routing_bias = nn.Parameter(
            torch.zeros(self.num_experts)
        )

        # ============================================================
        # ROUTED EXPERTS
        #
        # Mỗi expert là SwiGLU:
        #
        #                gate_proj → SiLU ─┐
        # x                              × → out_proj
        #                up_proj ─────────┘
        #
        # emb_dim -> hidden_dim -> emb_dim
        # ============================================================
        self.gate_proj = nn.ModuleList([
            nn.Linear(
                self.emb_dim,
                self.hidden_dim,
                bias=False
            )
            for _ in range(self.num_experts)
        ])

        self.up_proj = nn.ModuleList([
            nn.Linear(
                self.emb_dim,
                self.hidden_dim,
                bias=False
            )
            for _ in range(self.num_experts)
        ])

        self.out_proj = nn.ModuleList([
            nn.Linear(
                self.hidden_dim,
                self.emb_dim,
                bias=False
            )
            for _ in range(self.num_experts)
        ])

        # ============================================================
        # SHARED EXPERT
        #
        # Shared expert KHÔNG qua router.
        # Tất cả token đều chạy qua expert này.
        #
        # DeepSeek:
        #
        # routed experts
        #       +
        # shared expert
        #
        # => final MoE output
        # ============================================================
        shared_hidden_dim = cfg.get(
            "shared_expert_hidden_dim",
            self.hidden_dim
        )

        self.shared_gate_proj = nn.Linear(
            self.emb_dim,
            shared_hidden_dim,
            bias=False
        )

        self.shared_up_proj = nn.Linear(
            self.emb_dim,
            shared_hidden_dim,
            bias=False
        )

        self.shared_out_proj = nn.Linear(
            shared_hidden_dim,
            self.emb_dim,
            bias=False
        )

        self.silu = SiLU()

    def forward(self, x):
        # x:
        # (batch, seq_len, emb_dim)

        batch, seq_len, _ = x.shape

        # ============================================================
        # 1. SHARED EXPERT
        #
        # Mọi token đều đi qua đây.
        #
        # x:
        # (B, T, emb_dim)
        #
        # shared_hidden:
        # (B, T, shared_hidden_dim)
        #
        # shared_out:
        # (B, T, emb_dim)
        # ============================================================
        shared_hidden = (
            self.silu(self.shared_gate_proj(x))
            * self.shared_up_proj(x)
        )

        shared_out = self.shared_out_proj(
            shared_hidden
        )

        # ============================================================
        # 2. ROUTER SCORE
        #
        # raw_scores:
        # (B, T, num_experts)
        #
        # DeepSeek mới mặc định dùng:
        #
        # sqrt(softplus(score))
        #
        # thay vì softmax toàn bộ experts.
        # ============================================================
        raw_scores = self.gate(x.float())

        original_scores = torch.sqrt(
            torch.nn.functional.softplus(raw_scores)
        )

        # ============================================================
        # 3. ROUTING BIAS
        #
        # Bias CHỈ ảnh hưởng việc chọn expert.
        #
        # original_scores:
        # dùng làm weight thật
        #
        # selection_scores:
        # dùng để Top-K
        # ============================================================
        selection_scores = (
            original_scores
            + self.routing_bias
        )

        # ============================================================
        # 4. TOP-K EXPERT SELECTION
        #
        # Ví dụ:
        #
        # num_experts = 8
        # top_k = 2
        #
        # token "AI":
        #
        # -> Expert 2
        # -> Expert 6
        #
        # topk_indices:
        # (B, T, top_k)
        # ============================================================
        topk_indices = torch.topk(
            selection_scores,
            self.num_experts_per_tok,
            dim=-1
        ).indices

        # ============================================================
        # 5. LẤY WEIGHT TỪ ORIGINAL SCORE
        #
        # Không lấy score + bias.
        #
        # Đây là điểm DeepSeek làm khá hay:
        #
        # bias quyết định "chọn ai"
        # original score quyết định "đóng góp bao nhiêu"
        # ============================================================
        topk_weights = torch.gather(
            original_scores,
            dim=-1,
            index=topk_indices
        )

        # Normalize các expert được chọn.
        #
        # Ví dụ:
        #
        # E2 = 0.8
        # E6 = 0.4
        #
        # ->
        #
        # E2 = 0.667
        # E6 = 0.333
        topk_weights = (
            topk_weights
            / topk_weights.sum(
                dim=-1,
                keepdim=True
            )
        )

        # Cast trở lại dtype của x
        topk_weights = topk_weights.to(x.dtype)

        # ============================================================
        # 6. FLATTEN TOKENS
        #
        # (B, T, D)
        #
        # ->
        #
        # (B*T, D)
        #
        # Giúp gom các token đi cùng expert lại.
        # ============================================================
        x_flat = x.reshape(
            batch * seq_len,
            self.emb_dim
        )

        routed_out_flat = torch.zeros(
            batch * seq_len,
            self.emb_dim,
            device=x.device,
            dtype=x.dtype
        )

        topk_indices_flat = topk_indices.reshape(
            batch * seq_len,
            self.num_experts_per_tok
        )

        topk_weights_flat = topk_weights.reshape(
            batch * seq_len,
            self.num_experts_per_tok
        )

        # ============================================================
        # 7. CHỈ CHẠY EXPERT ĐƯỢC SỬ DỤNG
        #
        # Ví dụ:
        #
        # token0 -> E1, E4
        # token1 -> E4, E7
        # token2 -> E1, E2
        #
        # unique_experts:
        #
        # [E1, E2, E4, E7]
        #
        # Expert khác không chạy.
        # ============================================================
        unique_experts = torch.unique(
            topk_indices_flat
        )

        for expert_id_tensor in unique_experts:

            expert_id = int(
                expert_id_tensor.item()
            )

            # mask:
            #
            # token nào chọn expert này?
            #
            # shape:
            # (B*T, top_k)
            mask = (
                topk_indices_flat
                == expert_id
            )

            # Token nào có expert_id ở bất kỳ slot nào
            token_mask = mask.any(
                dim=-1
            )

            selected_idx = token_mask.nonzero(
                as_tuple=False
            ).squeeze(-1)

            if selected_idx.numel() == 0:
                continue

            # ========================================================
            # 8. LẤY TOKEN INPUT CHO EXPERT
            # ========================================================
            expert_input = x_flat.index_select(
                0,
                selected_idx
            )

            # ========================================================
            # 9. SwiGLU EXPERT
            #
            # DeepSeek:
            #
            # gate = w1(x)
            # up   = w3(x)
            #
            # hidden = SiLU(gate) * up
            #
            # output = w2(hidden)
            # ========================================================
            gate = self.gate_proj[
                expert_id
            ](expert_input)

            up = self.up_proj[
                expert_id
            ](expert_input)

            hidden = (
                self.silu(gate)
                * up
            )

            expert_out = self.out_proj[
                expert_id
            ](hidden)

            # ========================================================
            # 10. ROUTING WEIGHT CỦA EXPERT NÀY
            #
            # Ví dụ:
            #
            # token0:
            # E1 = 0.7
            # E4 = 0.3
            #
            # Khi đang xử lý E1:
            #
            # lấy đúng 0.7
            # ========================================================
            mask_selected = mask[
                selected_idx
            ]

            slot_indices = (
                mask_selected
                .int()
                .argmax(
                    dim=-1,
                    keepdim=True
                )
            )

            selected_weights = torch.gather(
                topk_weights_flat.index_select(
                    0,
                    selected_idx
                ),
                dim=-1,
                index=slot_indices
            ).squeeze(-1)

            # ========================================================
            # 11. WEIGHTED EXPERT OUTPUT
            #
            # output_token +=
            #
            # expert_output × routing_weight
            # ========================================================
            routed_out_flat.index_add_(
                0,
                selected_idx,
                expert_out
                * selected_weights.unsqueeze(-1)
            )

        # ============================================================
        # 12. RESTORE SHAPE
        #
        # (B*T, emb_dim)
        #
        # ->
        #
        # (B, T, emb_dim)
        # ============================================================
        routed_out = routed_out_flat.reshape(
            batch,
            seq_len,
            self.emb_dim
        )

        # ============================================================
        # 13. FINAL MoE OUTPUT
        #
        # DeepSeek-style:
        #
        # final =
        #     routed experts output
        #   + shared expert output
        #
        # Shared expert luôn active.
        # Routed experts chỉ top-k active.
        # ============================================================
        return routed_out + shared_out
```

```python
import torch

torch.manual_seed(42)

cfg = {
    "emb_dim": 4,
    "hidden_dim": 8,
    "num_experts": 3,
    "num_experts_per_tok": 2,
}

moe = MoEFeedForward(cfg)
moe.eval()

tokens = ["I", "love", "AI"]

x = torch.randn(1, 3, 4)

# ============================================================
# 1. Router
# ============================================================

scores = moe.gate(x)

topk_scores, topk_indices = torch.topk(
    scores,
    cfg["num_experts_per_tok"],
    dim=-1
)

topk_probs = torch.softmax(topk_scores, dim=-1)

output = moe(x)


# ============================================================
# PRINT DỄ ĐỌC
# ============================================================

print("\n==============================")
print("      MoE ROUTING FLOW")
print("==============================")

for i, token in enumerate(tokens):

    print(f"\nTOKEN {i}: '{token}'")
    print("-" * 30)

    # Input embedding
    print(
        "Input embedding:",
        [round(v, 3) for v in x[0, i].tolist()]
    )

    # Router score của tất cả expert
    print("\nRouter scores:")

    for expert_id in range(cfg["num_experts"]):
        print(
            f"  Expert {expert_id}: "
            f"{scores[0, i, expert_id].item():.3f}"
        )

    # Experts được chọn
    print("\nSelected experts:")

    for rank in range(cfg["num_experts_per_tok"]):

        expert_id = topk_indices[0, i, rank].item()
        probability = topk_probs[0, i, rank].item()

        print(
            f"  → Expert {expert_id}"
            f"   weight = {probability:.3f}"
            f" ({probability * 100:.1f}%)"
        )

    # Output cuối
    print(
        "\nFinal token output:",
        [round(v, 3) for v in output[0, i].tolist()]
    )


print("\n==============================")
print("SUMMARY")
print("==============================")

for i, token in enumerate(tokens):

    routes = []

    for rank in range(cfg["num_experts_per_tok"]):

        expert_id = topk_indices[0, i, rank].item()
        probability = topk_probs[0, i, rank].item()

        routes.append(
            f"E{expert_id}({probability * 100:.1f}%)"
        )

    print(
        f"{token:<5} → "
        + " + ".join(routes)
    )
```

```python
class TransformerBlock(nn.Module):
    def __init__(self, cfg):
        super().__init__()
        self.att = GroupedQueryAttention(
            d_in=cfg["emb_dim"],
            d_out=cfg["emb_dim"],
            num_heads=cfg["n_heads"],
            num_kv_groups=cfg["n_kv_groups"],
            dropout=cfg["drop_rate"],
            qkv_bias=cfg["qkv_bias"],
        )
        self.ff = MoEFeedForward(cfg) if cfg["num_experts"] > 0 else FeedForward(cfg)
        self.norm1 = LayerNorm(cfg["emb_dim"])
        self.norm2 = LayerNorm(cfg["emb_dim"])
        self.drop_shortcut = nn.Dropout(cfg["drop_rate"])

    def forward(self, x, use_cache=False):
        # Shortcut connection for attention block
        shortcut = x
        x = self.norm1(x)

        # x = self.att(x)   # Shape [batch_size, num_tokens, emb_size]
        ####################################################
        #  KV cache-related
        x = self.att(x, use_cache=use_cache)
        ####################################################

        x = self.drop_shortcut(x)
        x = x + shortcut  # Add the original input back

        # Shortcut connection for feed-forward block
        shortcut = x
        x = self.norm2(x)
        use_cuda = torch.cuda.is_available()
        if use_cuda:
            torch.cuda.synchronize()
            torch.cuda.reset_peak_memory_stats()
            base_mem = torch.cuda.memory_allocated()
        start = time.perf_counter()
        x = self.ff(x)
        if use_cuda:
            torch.cuda.synchronize()
            peak_mem = torch.cuda.max_memory_allocated()
            FFN_MEM_BYTES.append(peak_mem - base_mem)
        FFN_TIME_MS.append((time.perf_counter() - start) * 1000.0)
        x = self.drop_shortcut(x)
        x = x + shortcut  # Add the original input back

        return x
```

```python
class GPTModel(nn.Module):
    '''Small GPT language model using the MHA/KV-cache blocks above.'''
    def __init__(self, cfg):
        super().__init__()
        self.cfg = cfg
        self.token_embedding = nn.Embedding(cfg['vocab_size'], cfg['emb_dim'], dtype=cfg['dtype'])
        self.position_embedding = nn.Embedding(cfg['context_length'], cfg['emb_dim'], dtype=cfg['dtype'])
        self.dropout = nn.Dropout(cfg['drop_rate'])
        self.blocks = nn.ModuleList([TransformerBlock(cfg) for _ in range(cfg['n_layers'])])
        self.final_norm = LayerNorm(cfg['emb_dim'])
        self.lm_head = nn.Linear(cfg['emb_dim'], cfg['vocab_size'], bias=False, dtype=cfg['dtype'])
        self.lm_head.weight = self.token_embedding.weight

    def reset_cache(self):
        for block in self.blocks:
            block.att.reset_cache()

    def forward(self, input_ids, use_cache=False):
        if input_ids.ndim != 2 or input_ids.shape[1] == 0:
            raise ValueError('input_ids must have shape (batch, tokens) and be non-empty')
        batch, tokens = input_ids.shape
        if tokens > self.cfg['context_length']:
            raise ValueError('input is longer than context_length')
        if use_cache and self.blocks[0].att.cache_k is not None:
            past_tokens = self.blocks[0].att.cache_k.shape[-2]
        else:
            past_tokens = 0
        if past_tokens + tokens > self.cfg['context_length']:
            raise ValueError('cached input exceeds context_length; reset_cache() first')
        positions = torch.arange(past_tokens, past_tokens + tokens, device=input_ids.device)
        x = self.token_embedding(input_ids) + self.position_embedding(positions).unsqueeze(0)
        x = self.dropout(x)
        for block in self.blocks:
            x = block(x, use_cache=use_cache)
        return self.lm_head(self.final_norm(x))

    @torch.inference_mode()
    def generate(self, input_ids, max_new_tokens, use_cache=True):
        output = input_ids.clone()
        self.eval()
        self.reset_cache()
        if use_cache:
            logits = self(output[:, -self.cfg['context_length']:], use_cache=True)
            for _ in range(max_new_tokens):
                next_token = logits[:, -1].argmax(dim=-1, keepdim=True)
                output = torch.cat((output, next_token), dim=1)
                if output.shape[1] >= self.cfg['context_length']:
                    self.reset_cache()
                    logits = self(output[:, -self.cfg['context_length']:], use_cache=True)
                else:
                    logits = self(next_token, use_cache=True)
        else:
            for _ in range(max_new_tokens):
                logits = self(output[:, -self.cfg['context_length']:])
                output = torch.cat((output, logits[:, -1].argmax(dim=-1, keepdim=True)), dim=1)
        return output

cfg = {
    'vocab_size': 49152,
    'context_length': 8192,

    'emb_dim': 576,
    'hidden_dim': 256,   # ← không phải 4*576 = 2304

    'n_heads': 9,
    'n_kv_groups': 3,     # 3 Q heads share 1 KV head
                          # => n_kv_heads = 9 // 3 = 3
    'num_experts': 16,
    'num_experts_per_tok': 2,
    
    'n_layers': 30,       # ← SmolLM2-135M thật là 30 layers

    'drop_rate': 0.0,
    'qkv_bias': False,

    'dtype': torch.float32,  # để học/debug thì FP32 OK
    
}
torch.manual_seed(123)
model = GPTModel(cfg).eval()
prompt = torch.randint(0, cfg['vocab_size'], (1, 8))
full_logits = model(prompt)
model.reset_cache()
_ = model(prompt[:, :4], use_cache=True)
cached_logits = model(prompt[:, 4:], use_cache=True)
assert torch.allclose(full_logits[:, 4:], cached_logits, atol=1e-4, rtol=1e-4)
model.reset_cache()
uncached = model.generate(prompt, max_new_tokens=4, use_cache=False)
cached = model.generate(prompt, max_new_tokens=4, use_cache=True)
assert torch.equal(uncached, cached)
print('model ready; logits:', tuple(full_logits.shape), 'generation parity: OK')
```

```python
# ============================================================
# PARAMETER COUNT FOR MoE
# ============================================================

total_params = sum(p.numel() for p in model.parameters())

trainable_params = sum(
    p.numel()
    for p in model.parameters()
    if p.requires_grad
)

# ------------------------------------------------------------
# Routed experts
# ------------------------------------------------------------

all_routed_expert_params = 0
active_routed_expert_params = 0

# Shared experts
shared_expert_params = 0

for block in model.blocks:
    moe = block.ff

    # ========================================================
    # 1 ROUTED EXPERT
    # ========================================================

    one_expert_params = (
        sum(p.numel() for p in moe.gate_proj[0].parameters())
        + sum(p.numel() for p in moe.up_proj[0].parameters())
        + sum(p.numel() for p in moe.out_proj[0].parameters())
    )

    # Tất cả routed experts tồn tại trong layer
    all_routed_expert_params += (
        one_expert_params
        * moe.num_experts
    )

    # Chỉ Top-K expert được dùng/token
    active_routed_expert_params += (
        one_expert_params
        * moe.num_experts_per_tok
    )

    # ========================================================
    # SHARED EXPERT
    # luôn active
    # ========================================================

    shared_expert_params += (
        sum(p.numel() for p in moe.shared_gate_proj.parameters())
        + sum(p.numel() for p in moe.shared_up_proj.parameters())
        + sum(p.numel() for p in moe.shared_out_proj.parameters())
    )


# ============================================================
# EMBEDDING PARAMS
# ============================================================

embedding_params = 0

for module in model.modules():
    if isinstance(module, torch.nn.Embedding):
        embedding_params += sum(
            p.numel()
            for p in module.parameters(recurse=False)
        )


# ============================================================
# ROUTER PARAMS
# ============================================================

router_params = 0

for block in model.blocks:
    moe = block.ff

    router_params += sum(
        p.numel()
        for p in moe.gate.parameters()
    )

    # routing_bias
    router_params += moe.routing_bias.numel()


# ============================================================
# OTHER DENSE PARAMS
#
# attention, norm, LM head, ...
#
# Chú ý:
# shared expert và router đã nằm trong total_params.
# ============================================================

other_dense_params = (
    total_params
    - all_routed_expert_params
    - shared_expert_params
    - router_params
    - embedding_params
)


# ============================================================
# ACTIVATED PARAMS
#
# Kiểu "A3B":
#
# dense non-embedding
# + shared experts
# + router
# + Top-K routed experts
# ============================================================

active_non_embedding_params = (
    other_dense_params
    + router_params
    + shared_expert_params
    + active_routed_expert_params
)


# ============================================================
# PRINT
# ============================================================

print("=" * 65)
print("MODEL PARAMETER REPORT")
print("=" * 65)

print(f"Total params:                  {total_params:>15,}")
print(f"Trainable params:              {trainable_params:>15,}")
print()

print("-" * 65)
print("ROUTED EXPERTS")
print("-" * 65)

print(f"Total routed expert params:    {all_routed_expert_params:>15,}")
print(f"Active routed params/token:    {active_routed_expert_params:>15,}")
print()

print("-" * 65)
print("OTHER COMPONENTS")
print("-" * 65)

print(f"Shared expert params:          {shared_expert_params:>15,}")
print(f"Router params:                 {router_params:>15,}")
print(f"Embedding params:              {embedding_params:>15,}")
print(f"Other dense params:            {other_dense_params:>15,}")
print()

print("-" * 65)
print("MODEL SIZE")
print("-" * 65)

print(f"Total size:                    {total_params / 1e9:>14.3f} B")
print(
    f"Active size (non-embedding):   "
    f"{active_non_embedding_params / 1e9:>14.3f} B"
)

print(
    f"Active ratio:                  "
    f"{active_non_embedding_params / total_params * 100:>13.2f} %"
)

print()

print("-" * 65)
print("MoE CONFIG")
print("-" * 65)

moe = model.blocks[0].ff

print(f"Number of layers:              {len(model.blocks):>15}")
print(f"Experts / layer:               {moe.num_experts:>15}")
print(f"Active experts / token/layer:  {moe.num_experts_per_tok:>15}")
print(f"Params / routed expert:        {one_expert_params:>15,}")

print(
    f"Expert activations/token:      "
    f"{len(model.blocks) * moe.num_experts_per_tok:>15}"
)

print("=" * 65)
```
